```python
import pandas as pd

# Read in file
df = pd.read_csv('data/wetlands_seasonal_bird_diversity.csv')

# Check the first 5 rows
df.head()
```

```python
# Default plot(): one line per column with numeric data
df.plot()
```

# Line plots

```python
# Bird species registered during winter at CSM yearly
df.plot(x='year', y = 'CSM_winter')
```

```python
# Customization

df.plot(x='year',
       y = 'CSM_winter',
       title = 'Bird species registered during winter at Carpinteria Salt Marsh',
       xlabel = 'year',
       ylabel = 'Number of bird species',
       color = 'green',
       legend = False)
```

```python
## Check-in
df.plot(x='year',
       y = 'MUL_winter',
       title = 'Bird species registered during winter at Mugu Lagoon',
       xlabel = 'year',
       ylabel = 'Number of bird species',
       color = 'blue',
       legend = False)
```

```python
df.plot(x='year',
       y = 'MUL_spring',
       title = 'Bird species registered during spring at Mugu Lagoon',
       xlabel = 'year',
       ylabel = 'Number of bird species',
       color = 'purple',
       legend = False)
```

```python
df.plot(x='year',
       y = 'MUL_fall',
       title = 'Bird species registered during fall at Mugu Lagoon',
       xlabel = 'year',
       ylabel = 'Number of bird species',
       color = 'orange',
       legend = False)
```

```python
df[df['MUL_spring'].isna()]
```

# Multiple line plots

```python
# Plot the bird surveys at the Tijuana Estuary during spring and fall across years

df.plot(x = 'year',
       y = ['TJE_spring', 'TJE_fall'],
       title = 'Seasonal bird surveys at Tijuana Estuary',
       xlabel = 'Year',
       ylabel = 'Number of bird species',
       color = {'TJE_spring': '#F48FB1',
               'TJE_fall': '#AB47BC'}
       )
```

```python
# Create seperate plots for each column by setting the subplots parameter to True

df.plot(x = 'year',
       y = ['TJE_spring', 'TJE_fall'],
       title = 'Seasonal bird surveys at Tijuana Estuary',
       xlabel = 'Year',
       ylabel = 'Number of bird species',
       color = {'TJE_spring': '#F48FB1',
               'TJE_fall': '#AB47BC'},
        subplots = True
       )
```

# Updating the index

```python
# Set `column_name` column in df as the new index 
df.set_index('year')
df.head()
```

```python
# Simple plot of Carpinteria Salt Marsh winter surveys
df.plot(y = 'CSM_winter')
```

```python
#  You can reset the index 
df = df.reset_index()
df.head()
```

## Check-in

1. `df.set_index('year')` is setting year column as the index of df. `.loc[:,'SDW_winter':'TJE_fall']` is selecting the all the rows and the columns from `'SDW_winter'` to `'TJE_spring'`. `plot()` is plotting all those columns onto the same plot.
2. Yes, the code modifies the df, it is changing the index.
3. It would make more sense to compare the surveys between season per site.

```python
df.set_index('year').loc[:,'SDW_winter':'TJE_fall'].plot()
```

# Method chaining

```python
(df.set_index('year')
    .loc[:, 'SDW_winter':'TJE_fall']
.plot()
)
```

```python
year_index_df = df.set_index('year')
subset_df = year_index_df.loc[:, 'SDW_winter':'TJE_fall']
subset_df.plot()
```

# Data exploration

```python
# Read in data
URL = 'https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv'
penguins = pd.read_csv(URL)

penguins.head()
```

```python
# Check columnn data types and NA values
penguins.info()
```

```python
# Simple statistics about numeric colums
penguins.describe()
```

```python
# Count unique values in categorical columns and year
penguins[['species', 'island', 'sex', 'year']].nunique()
```

```python
penguins['species'].unique()
```

```python
# Number of values per unique value in species column
penguins['species'].value_counts()
```

# `kind` argument in `plot()`

the default of `kind` is `line`

# Scatter plots

```python
penguins.plot(kind = 'scatter',
             x = 'flipper_length_mm',
             y = 'body_mass_g')
```

```python
penguins.plot(kind = 'scatter',
             x = 'flipper_length_mm',
             y = 'body_mass_g',
             title = 'Flipper length and body mass per Palmer penguins',
             xlabel = 'Flipper length (mm)',
             ylabel = 'Body mass (g)',
             color = '#ff3b01',
             alpha = 0.4 # Controls transparency
             )
```

# Bar plots

```python
smallest = penguins['body_mass_g'].nsmallest(10)
smallest
```

```python
smallest.plot(kind = 'bar')
```

```python
penguins.nsmallest(10, 'body_mass_g')
```

# Histograms

```python
# Using plot without subsetting data - a mess again 
penguins.plot(kind = 'hist')
```

```python
# Subset before plotting
# Distribution of flipper length measurements
# First select data, then plot
penguins['flipper_length_mm'].plot(kind = 'hist',
                                  title = 'Penguin flipper lengths',
                                  xlabel = 'Flipper length (mm)',
                                  grid = True)
```

## Check-in

```python
# Boxplots of penguin bill length and depth
penguins[['bill_length_mm', 'bill_depth_mm']].plot(kind = 'box',
                                                  ylabel = 'Length (mm)',
                                                  title = 'Penguin bill length and depth')
```

```python
female_gentoo = penguins[(penguins['sex'] == 'female') & (penguins['species'] == 'Gentoo')]
female_gentoo.head()
```

```python
female_gentoo['flipper_length_mm'].plot(kind = 'hist',
                                       xlabel = 'Flipper length (mm)',
                                       title = 'Flipper length (mm) of female Gentoo penguins')
```
